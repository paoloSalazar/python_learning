## Closures


Closures functions

```python
def power(exponent):
    def inner(base):
        return base ** exponent
    return inner

power_of_two = power(2)
power_of_three = power(3)

print(power_of_two(5))  # Output: 25
print(power_of_three(5))  # Output: 8
```

Python creates something named NONLOCAL Variables 
Exponent is the so-called free non-local variable, and it's keept in memory even after the local scope of the enclosing function is gone

A function that utilizes non-local variables is known as closure because it encloses the needed nonlocal variables together with the function object
A closure is in a way hiding the non local variables from the global scope and other functions because nonlocal variables are only available from inside of the closure and nowhere else
The closure will package a tuple of so-called cell variables with the function object.

### Decorating Functions
The point of a decorator function is to create a new inner function which will add some additional behavior

```python
def email_decorator(func):
    ### The Wraper function is a closure because it is callingthe func function defined outside its scope
    def wrapper():
        print("Dear Interns,")
        func()
        print("Best REgards")
        print("Your new Boss")
    return wrapper

## sintax sugar to apply a decorator to a function
@email_decorator
def greeting_message():
    print("Welcome to your new job!")

### Of course we could have assigned the decorated function to a new variable
### but the point of decorators is to use the same function with additional behavior
### without actually modifying the function itself

# greeting_message = email_decorator(greeting_message)

greeting_message()
```
